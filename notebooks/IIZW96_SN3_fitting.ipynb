```python
from scipy.ndimage import gaussian_filter  # optional if you want smoothing

def show_sigma_map(cube_, path_sigma, path_flux, path_flux_err,
                   smooth_sigma=None):
    fig = plt.figure()
    ax = fig.add_subplot(111, projection=cube_.get_wcs())

    # Axis formatting
    ax.coords[0].set_major_formatter('hh:mm:ss.s')
    ax.coords[1].set_major_formatter('dd:mm:ss')
    ax.coords[0].set_axislabel('R.A. (J2000)')
    ax.coords[1].set_axislabel('Dec. (J2000)', minpad=-0.5)

    # Read maps
    sigma_map = orb.utils.io.read_fits(path_sigma)      # km/s (usually)
    flux_map = orb.utils.io.read_fits(path_flux)
    flux_err_map = orb.utils.io.read_fits(path_flux_err)

    # S/N masking using flux
    line_snr = flux_map / flux_err_map
    sigma_map[line_snr <= 3] = np.nan  # keep only >3σ detections

    # Optional smoothing in pixel space
    if smooth_sigma is not None:
        sigma_map = gaussian_filter(sigma_map, sigma=smooth_sigma)

    # Choose display limits (you can tweak these)
    vmin = np.nanpercentile(sigma_map, 5)
    vmax = np.nanpercentile(sigma_map, 95)
    print("sigma range (km/s):", np.nanmin(sigma_map), np.nanmax(sigma_map))

    im = ax.imshow(
        sigma_map.T,
        origin='lower',
        cmap='magma',      # sequential is better for σ
        vmin=vmin,
        vmax=vmax
    )

    # Region markers
    ax.plot_coord(position_ID8, "x", c='black', label='8/D1')
    ax.plot_coord(position_ID7, "x", c='c', label='7/D0')
    ax.plot_coord(position_ID11, "x", c='salmon', label='11/C0')
    ax.legend(loc='upper left')

    # Zoom (same as your other maps)
    ax.set_xlim(1000, 1110)
    ax.set_ylim(1000, 1110)

    # Colorbar
    cbar = plt.colorbar(im, ax=ax, pad=0.02)
    cbar.set_label(r"$\sigma$ (km s$^{-1}$)")

    #plt.savefig("/Users/leianydeoleo/Desktop/IIZw96_broadening_SN3.pdf"),
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
from astropy.wcs import WCS

flux_file = "/Users/leianydeoleo/IIZw96_SN3/MAPS/IIZw96_SN3.LineMaps.map.6563.1x1.flux.fits"
vel_file = "/Users/leianydeoleo/IIZw96_SN3/MAPS/IIZw96_SN3.LineMaps.map.all.1x1.velocity.fits"
err_file = "/Users/leianydeoleo/IIZw96_SN3/MAPS/IIZw96_SN3.LineMaps.map.6563.1x1.flux-err.fits"

flux_data, header = fits.getdata(flux_file, header=True)
vel_data = fits.getdata(vel_file)
flux_err = fits.getdata(err_file)

def plot_flux_with_velocity_contours_wcs(
    flux_data,
    vel_data,
    header,
    flux_err=None,
    title="Hα Flux with Velocity Contours",
    cmap_flux="inferno",
    levels=None,
    vmin=None,
    vmax=None,
    threshold_sigma=1.0
):
    wcs = WCS(header)

    fig = plt.figure(figsize=(8, 6))
    ax = plt.subplot(projection=wcs)

    ax.coords[0].set_major_formatter('hh:mm:ss.s')
    ax.coords[1].set_major_formatter('dd:mm:ss')
    ax.set_xlabel("R.A. (J2000)")
    ax.set_ylabel("Dec. (J2000)")

    im = ax.imshow(
        flux_data,
        origin="lower",
        cmap=cmap_flux,
        vmin=vmin if vmin is not None else np.nanpercentile(flux_data, 0),
        vmax=vmax if vmax is not None else np.nanpercentile(flux_data, 99)
    )

    plt.colorbar(im, ax=ax, label="Hα Flux")

    if flux_err is None:
        raise ValueError("flux_err is required to build the flux threshold mask.")

    sigma_flux = np.nanmedian(flux_err)
    flux_mask = flux_data > (threshold_sigma * sigma_flux)

    vel_clean = np.ma.masked_where(~flux_mask, vel_data)

    if vel_clean.count() == 0:
        raise ValueError("No unmasked velocity pixels remain. Lower threshold_sigma.")

    vel_values = vel_clean.compressed()

    if levels is None:
        mean_vel = np.nanmean(vel_values)
        std_vel = np.nanstd(vel_values)
        levels = np.linspace(mean_vel - 2 * std_vel, mean_vel + 2 * std_vel, 50)
       

    cs = ax.contour(
    vel_clean,
    levels=levels,
    colors="cyan",
    linewidths=1,
    zorder=5
     )
    
    ax.clabel(
    cs,
    cs.levels,
    fmt=lambda v: f"{v:.0f} km/s",
    fontsize=8,
    colors="white",
    inline=True,
    manual=[(1020, 1020), (1060, 1060), (1090, 1070), (1080,1060), (1074.91441947,1033.22775456)]
    )

    ax.set_title(title)
    ax.set_xlim(1000, 1110)
    ax.set_ylim(1000, 1110)

    plt.tight_layout()
    plt.show()

plot_flux_with_velocity_contours_wcs(
    flux_data,
    vel_data,
    header,
    flux_err=flux_err,
    threshold_sigma=15.0
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.visualization import ImageNormalize, LogStretch
from scipy.ndimage import gaussian_filter


def make_Ha_flux_with_flux_contours(
    flux_map,
    wcs,
    title=r"H$\alpha$ collapsed emission contours",
    smooth_sigma_image=0.5,
    smooth_sigma_contour=0.5,
    x_range_pix=(1000, 1120),
    y_range_pix=(1000, 1100),
    #sourceD_pix=(x_pix, y_pix),
    vmin_fixed=1e-18,        # <-- set your colorbar lower bound here
    vmax_fixed=1e-15,
    savepath=None
):
    """Hα flux map (log) with Hα flux contours (DS9-style)."""

    cel_wcs = wcs.celestial

    # --- base image (smooth, then log) ---
    img = np.array(flux_map, dtype=float)
    img[~np.isfinite(img)] = np.nan

    if smooth_sigma_image is not None:
        img = gaussian_filter(img, smooth_sigma_image)

    img[img <= 0] = np.nan
    finite = np.isfinite(img)

    vmin = np.nanpercentile(img[finite], 5)
    vmax = np.nanpercentile(img[finite], 99.5)

    norm = ImageNormalize(vmin=vmin, vmax=vmax, stretch=LogStretch())

    fig = plt.figure(figsize=(7.5, 5.5))
    ax = fig.add_subplot(111, projection=cel_wcs)

    im = ax.imshow(img, origin="lower", cmap="nipy_spectral", norm=norm)

    # --- flux contours (same map, slightly different smoothing) ---
    cdata = np.array(flux_map, dtype=float)
    cdata[~np.isfinite(cdata)] = np.nan
    if smooth_sigma_contour is not None:
        cdata = gaussian_filter(cdata, smooth_sigma_contour)
    cdata[cdata <= 0] = np.nan

    cfinite = np.isfinite(cdata)
    cmin = np.nanpercentile(cdata[cfinite], 40)
    cmax = np.nanpercentile(cdata[cfinite], 99.7)
    levels = np.geomspace(cmin, cmax, 12)

    ax.contour(cdata, levels=levels,
               colors="black", linewidths=0.8, origin="lower")


    # --- zoom ---
    if x_range_pix is not None:
        ax.set_xlim(x_range_pix)
    if y_range_pix is not None:
        ax.set_ylim(y_range_pix)

    # --- titles and labels ---
    ax.coords[0].set_axislabel("Right Ascension (J2000)")
    ax.coords[1].set_axislabel("Declination (J2000)")
    ax.set_title("Collapsed Halpha emission over Flux")
    # --- colorbar with fixed size ---
    # Add a smaller, proportional colorbar next to the plot
    # --- DS9-style colorbar aligned with map ---
    cbar = fig.colorbar(im)
    cbar.set_label(r"H$\alpha$ Flux (erg s$^{-1}$ cm$^{-2}$)", fontsize=11)
    cbar.ax.tick_params(labelsize=9, width=0.8, length=3)

# Show full range of values (log spaced)
    ticks = np.logspace(np.log10(vmin_fixed), np.log10(vmax_fixed), num=8)
    cbar.set_ticks(ticks)
    cbar.set_ticklabels([f"{t:.0e}" for t in ticks])

    plt.show()
    return fig, ax


# ======== EXAMPLE CALL ========

flux_map = flux_map
wcs = cube_SN3.get_wcs()

fig, ax = make_Ha_flux_with_flux_contours(
    flux_map.T,
    wcs,
    # optional zoom:
    # x_range_pix=(1000, 1120),
    # y_range_pix=(1000, 1100),
    savepath="/Users/leianydeoleo/Desktop/IIZw96_Ha_flux_log_fluxcontours_final.png"
)

#fig.savefig(
#    "/Users/leianydeoleo/Desktop/IIZw96_Ha_flux_log_contour.png",
#  dpi=300,
#    bbox_inches="tight")
```
