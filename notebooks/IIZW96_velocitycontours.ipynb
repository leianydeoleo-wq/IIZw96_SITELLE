```python
flux_file = "/Users/leianydeoleo/Desktop/IIZW96/IIZw96_SN3/MAPS/IIZw96_SN3.LineMaps.map.6563.1x1.flux.fits"
vel_file = "/Users/leianydeoleo/Desktop/IIZW96/IIZw96_SN3/MAPS/IIZw96_SN3.LineMaps.map.all.1x1.velocity.fits"
err_file = "/Users/leianydeoleo/Desktop/IIZW96/IIZw96_SN3/MAPS/IIZw96_SN3.LineMaps.map.6563.1x1.flux-err.fits"

flux_data, header = fits.getdata(flux_file, header=True)
vel_data = fits.getdata(vel_file)
flux_err = fits.getdata(err_file)

def plot_flux_with_velocity_contours_wcs(
    flux_data,
    vel_data,
    header,
    flux_err=None,
    title="Hα Flux with Velocity Contours",
    cmap_flux="inferno",
    levels=None,
    vmin=None,
    vmax=None,
    threshold_sigma=1.0
):
    wcs = WCS(header)

    fig = plt.figure(figsize=(8, 6))
    ax = plt.subplot(projection=wcs)

    ax.coords[0].set_major_formatter('hh:mm:ss.s')
    ax.coords[1].set_major_formatter('dd:mm:ss')
    ax.set_xlabel("R.A. (J2000)")
    ax.set_ylabel("Dec. (J2000)")

    im = ax.imshow(
        flux_data,
        origin="lower",
        cmap=cmap_flux,
        vmin=vmin if vmin is not None else np.nanpercentile(flux_data, 0),
        vmax=vmax if vmax is not None else np.nanpercentile(flux_data, 99)
    )

    plt.colorbar(im, ax=ax, label="Hα Flux")

    if flux_err is None:
        raise ValueError("flux_err is required to build the flux threshold mask.")

    sigma_flux = np.nanmedian(flux_err)
    flux_mask = flux_data > (threshold_sigma * sigma_flux)

    vel_clean = np.ma.masked_where(~flux_mask, vel_data)

    if vel_clean.count() == 0:
        raise ValueError("No unmasked velocity pixels remain. Lower threshold_sigma.")

    vel_values = vel_clean.compressed()

    if levels is None:
        mean_vel = np.nanmean(vel_values)
        std_vel = np.nanstd(vel_values)
        levels = np.linspace(mean_vel - 2 * std_vel, mean_vel + 2 * std_vel, 100)
       

    cs = ax.contour(
    vel_clean,
    levels=levels,
    colors="cyan",
    linewidths=1,
    zorder=5
     )
    
    ax.clabel(
    cs,
    cs.levels,
    fmt=lambda v: f"{v:.0f} km/s",
    fontsize=8,
    colors="white",
    inline=True,
    manual=[(1020, 1020), (1060, 1060), (1090, 1070), (1080,1060), (1074.91441947,1033.22775456)]
    )

    ax.set_title(title)
    ax.set_xlim(1000, 1110)
    ax.set_ylim(1000, 1110)
    plt.savefig("/Users/leianydeoleo/Desktop/IIZw96_hacontours.pdf")

    plt.tight_layout()
    plt.show()

plot_flux_with_velocity_contours_wcs(
    flux_data,
    vel_data,
    header,
    flux_err=flux_err,
    threshold_sigma=15.0)
```
